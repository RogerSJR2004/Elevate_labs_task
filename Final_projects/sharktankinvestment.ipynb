```python
# Clean data (basic)
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.drop_duplicates(inplace=True)
df['deal'] = df['deal'].astype(bool)
```

# Analyze funding trends

```python
# Assuming 'industry_deals' is already calculated as in your original code
industry_deals = df['idea'].value_counts().head(10)

# Create a pie chart without percentage values inside slices
plt.figure(figsize=(10, 10))
plt.pie(industry_deals.values, labels=industry_deals.index, startangle=90,
        colors=sns.color_palette('pastel'))  # Removed autopct
plt.title('Top 10 Startup Ideas / Industries', fontsize=16)

# Add a legend with percentage values
plt.legend(industry_deals.index, loc="center left", bbox_to_anchor=(1, 0, 0.5, 1),
           title="Industry (Percentage)",  # Add title to legend
           labels=['%s (%.1f%%)' % (l, s * 100 / industry_deals.sum()) for l, s in zip(industry_deals.index, industry_deals.values)])  # Format legend labels with percentages

plt.tight_layout()
plt.show()
```

```python
## Industry-wise deal count
industry_deals = df['idea'].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=industry_deals.values, y=industry_deals.index, palette="crest")
plt.title("Top 10 Startup Ideas / Industries")
plt.xlabel("Number of Startups")
plt.ylabel("Industry")
plt.tight_layout()
plt.show()
```

```python
## Total funding by industry (idea)
industry_funding = df.groupby('idea')['deal_amount'].sum().sort_values(ascending=False).head(10)
industry_funding.plot(kind='barh', title="Top Funded Startup Domains")
plt.xlabel("Total Deal Amount (in Lakhs)")
plt.tight_layout()
plt.show()
```

```python
## Average equity taken per deal
plt.figure(figsize=(8,5))
sns.histplot(df[df["deal"] == True]['deal_equity'], bins=15, kde=True, color="orange")
plt.title("Equity Distribution for Funded Startups")
plt.xlabel("Equity (%)")
plt.tight_layout()
plt.show()
```

```python
# Shark Participation

## Number of deals by each shark
sharks = ['ashneer_deal', 'anupam_deal', 'aman_deal', 'namita_deal',
          'vineeta_deal', 'peyush_deal', 'ghazal_deal']
shark_counts = df[sharks].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
sns.barplot(x=shark_counts.index.str.replace('_deal', '').str.title(), y=shark_counts.values, palette='Set2')
plt.title("Number of Deals by Each Shark")
plt.ylabel("Deals")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Success Rate (Founders who got a deal)
success_rate = df['deal'].mean() * 100
print(f"✅ Overall Deal Success Rate: {success_rate:.2f}%")
```

```python
# Funding stage profile: Equity vs Amount
plt.figure(figsize=(8,6))
sns.scatterplot(data=df[df["deal"] == True], x="deal_amount", y="deal_equity", hue="total_sharks_invested", palette="cool", size="total_sharks_invested")
plt.title("Deal Amount vs Equity Given (colored by number of sharks)")
plt.xlabel("Deal Amount (Lakhs)")
plt.ylabel("Equity (%)")
plt.tight_layout()
plt.show()
```
